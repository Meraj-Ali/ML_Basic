1.Use iris flower dataset from sklearn library and try to form clusters of flowers using petal width and length features. Drop other two features for simplicity.

2.Figure out if any preprocessing such as scaling would help here

3.Draw elbow plot and from that figure out optimal value of k

```python
from sklearn.cluster import KMeans
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('iris.csv')
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
plt.scatter(df['SepalLengthCm'],df['Species'])
plt.scatter(df['SepalWidthCm'],df['Species'])
plt.scatter(df['PetalLengthCm'],df['Species'])
plt.scatter(df['SepalWidthCm'],df['Species'])
plt.xlabel('Lengthcm')
plt.ylabel('Species')
```

```python
km = KMeans(n_clusters=3)
y_predicted = km.fit_predict(df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']])
y_predicted
```

```python
df['cluster'] = y_predicted
df.head()
```

```python
km.cluster_centers_
```

```python
df1 = df[df.cluster==0]
df2 = df[df.cluster==1]
df3 = df[df.cluster==2]
```

```python
plt.scatter(df1.Species, df1['SepalLengthCm'], color='green')
plt.scatter(df1.Species, df1['SepalWidthCm'], color='red')
plt.scatter(df1.Species, df1['PetalLengthCm'], color='black')
plt.scatter(df1.Species, df1['PetalWidthCm'], color='yellow')
plt.scatter(df2.Species, df2['SepalLengthCm'], color='green')
plt.scatter(df2.Species, df2['SepalWidthCm'], color='red')
plt.scatter(df2.Species, df2['PetalLengthCm'], color='black')
plt.scatter(df2.Species, df2['PetalWidthCm'], color='yellow')
plt.scatter(df3.Species, df3['SepalLengthCm'], color='green')
plt.scatter(df3.Species, df3['SepalWidthCm'], color='red')
plt.scatter(df3.Species, df3['PetalLengthCm'], color='black')
plt.scatter(df3.Species, df3['PetalWidthCm'], color='yellow')
plt.xlabel('Species')
plt.ylabel('LengthCm')
plt.legend()
```
