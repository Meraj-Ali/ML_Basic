```python
import pandas as pd
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df.drop(['PassengerId', 'Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns', inplace=True)
df.head()
```

```python
target=df.Survived
inputs=df.drop('Survived',axis='columns')
```

```python
# one hot encoding 
dummies = pd.get_dummies(inputs.Sex, dtype=int)
dummies.head(3)
```

```python
# concat inputs and dummies 
inputs = pd.concat([inputs, dummies], axis='columns')
inputs.head(3)
```

```python
# drop Sex columns
inputs.drop('Sex', axis='columns', inplace=True)
inputs.head(3)
```

```python
# find nan number in any columns
inputs.columns[inputs.isna().any()]
```

```python
inputs.Age[:10]
```

```python
# use mean to fill those nan values
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head(6)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(inputs, target, test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
len(inputs)
```

```python
# create naive bayes model
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
```

```python
# train the model
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

# Useing Spam datasets 

```python
# Spam datasets 
import pandas as pd
df = pd.read_csv('spam_ham_dataset.csv')
df.head()
```

```python
df.groupby('label').describe()
```

```python
# label column changinh binary 
df['spam']=df['label'].apply(lambda x: 1 if x=='spam' else 0)
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.text, df.spam, test_size=0.25)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
v = CountVectorizer()
X_train_count = v.fit_transform(X_train.values)
X_train_count.toarray()[:3]
```

```python
# use multinomialNB 
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_count, y_train)
```

```python
emails = [
    'hpl nom for january 9 , 2001\r\n(see...',
    'indian springs\r\nthis deal is t...'
]
emails_count = v.transform(emails)
model.predict(emails_count)
```

```python
X_test_count = v.transform(X_test)
model.score(X_test_count,y_test)
```

```python
# useing sklearn pipeline 
from sklearn.pipeline import Pipeline
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])
```

```python
clf.fit(X_train, y_train)
```

```python
clf.score(X_test, y_test)
```
